# Exploratory Data Analysis

## 1. Target Distribution
Check for class imbalance in `Cover_Type`.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load data
df = pd.read_csv('../data/train.csv')

# Countplot
plt.figure(figsize=(10, 6))
sns.countplot(data=df, x='Cover_Type')
plt.title('Distribution of Cover_Type')
plt.show()
```

## 2. Topographic Analysis
Boxplots of `Elevation`, `Aspect`, `Slope` grouped by `Cover_Type`.

```python
topo_cols = ['Elevation', 'Aspect', 'Slope']

plt.figure(figsize=(15, 5))
for i, col in enumerate(topo_cols):
    plt.subplot(1, 3, i+1)
    sns.boxplot(data=df, x='Cover_Type', y=col)
    plt.title(f'{col} by Cover_Type')
plt.tight_layout()
plt.show()
```

## 3. Correlation Heatmap
Heatmap of numerical features.

```python
num_cols = ['Elevation', 'Aspect', 'Slope', 'Horizontal_Distance_To_Hydrology', 
            'Vertical_Distance_To_Hydrology', 'Horizontal_Distance_To_Roadways',
            'Hillshade_9am', 'Hillshade_Noon', 'Hillshade_3pm',
            'Horizontal_Distance_To_Fire_Points']

plt.figure(figsize=(12, 10))
sns.heatmap(df[num_cols].corr(), annot=True, fmt='.2f', cmap='coolwarm')
plt.title('Correlation Matrix of Numerical Features')
plt.show()
```

## 4. Wilderness/Soil Analysis
Bar charts showing frequency of binary columns.

```python
wild_cols = [c for c in df.columns if 'Wilderness_Area' in c]
soil_cols = [c for c in df.columns if 'Soil_Type' in c]

# Wilderness
wild_counts = df[wild_cols].sum()
plt.figure(figsize=(10, 6))
sns.barplot(x=wild_counts.index, y=wild_counts.values)
plt.xticks(rotation=45)
plt.title('Wilderness Area Frequency')
plt.show()

# Soil Types (Top 10 to avoid clutter)
soil_counts = df[soil_cols].sum().sort_values(ascending=False).head(10)
plt.figure(figsize=(12, 6))
sns.barplot(x=soil_counts.index, y=soil_counts.values)
plt.xticks(rotation=45)
plt.title('Top 10 Soil Types Frequency')
plt.show()
```

## 5. Conclusion

### Key Insights:
1. **Class Imbalance:** The target variable `Cover_Type` shows a significant imbalance, with some classes (e.g., 1 and 2) being much more frequent than others. This suggests that using stratified sampling and potentially class weighting in models (like in LightGBM) is crucial.
2. **Elevation is a Strong Predictor:** The boxplots reveal distinct elevation ranges for different cover types, indicating that `Elevation` is likely the most important feature for classification.
3. **Soil Type Diversity:** There are many soil types, some of which are rare. Grouping them into Climatic and Geologic zones, as done in the feature engineering step, should help reduce dimensionality and capture broader patterns.
